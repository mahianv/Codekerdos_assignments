```python
## 23. Check whether a list contains a sublist
## A sublist means consecutive elements in the same order.
def has_sublist(main, sub):
    for i in range(len(main) - len(sub) + 1):
        if main[i:i + len(sub)] == sub:
            return True
    return False

(has_sublist([1, 2, 3, 4, 5], [3, 4]), has_sublist([1, 2, 3, 4, 5], [2, 4]))
```

```python
## 24. Generate all sublists of a list
def all_sublists(lst):
    result = [[]]
    for i in range(len(lst)):
        for j in range(i + 1, len(lst) + 1):
            result.append(lst[i:j])
    return result

all_sublists([1, 2, 3]) 
```

```python
## 26. How to flatten a list in Python?
def flatten(lst):
    flat = []
    for item in lst:
        if isinstance(item, list):
            flat.extend(flatten(item))
        else:
            flat.append(item)
    return flat

flatten([1, [2, 3], [4, [5, 6]], 7]) 
```

```python
## 38. Replace each element with the next larger number to its right (else -1)
nums = [4, 2, 6, 8, 1, 2]
out = []

for i in range(len(nums)):
    ans = -1
    for j in range(i + 1, len(nums)):
        if nums[j] > nums[i]:
            ans = nums[j]
            break
    out.append(ans)

out
```
